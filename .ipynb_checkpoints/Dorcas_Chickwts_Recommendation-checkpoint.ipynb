# MY BUSINESS QUESTION:
For a chicken with a particular target weight, which feed should we consider?
We shall build a recommendation function theat will help in answering our question or will offer suggestion.

```python
# import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# loading chickwts dataset
df = pd.read_csv('chickwts.csv')
df.head()
```

```python
# looking at the rows and columns
df.shape
```

we have 71 rows and 2 columns

```python
df.info()
```

```python
# checking for missing values
df.isnull().sum()
```

found no missing value

```python
print("\nFeed types:")
print(df["feed"].unique())
```

```python
print("\nNumber of chickens per feed:")
print(df["feed"].value_counts())
```

```python
# Understanding the performance of each feed
feed_summary = df.groupby('feed')['weight'].agg(
    ['count','mean','min','max']
).sort_values('count',ascending=False)

feed_summary
```

count:	Number of chickens receiving that feed

mean:	Average chicken weight

min:	    Smallest recorded weight

max:	    Largest recorded weight

```python
# visualizing feed_summary
feed_summary["count"].sort_values().plot(
    kind="barh",
    figsize=(8, 5)
)

plt.xlabel("Total Chicken Weight")
plt.ylabel("Feed Type")
plt.title("Total Chicken Weight by Feed Type")
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(data=df, x="feed", y="weight")

plt.xlabel("Feed Type")
plt.ylabel("Chicken Weight")
plt.title("Chicken Weight Distribution by Feed Type")
plt.xticks(rotation=45)
plt.show()
```

Horsebean has the lowest chicken weight.

Casein has the highest median weight.

Sunflower also has relatively high weights, but it has a few unusual values (outliers).

Meatmeal has a fairly wide spread of weights.

Linseed and soybean fall between the lower and higher groups.

There is variation within every feed type, so we shouldn't say that one feed causes higher weight just from the dataset.

```python
feed_profile = df.groupby('feed')['weight'].agg(
    mean='mean',
    median='median',
    std='std',
    min='min',
    max='max',
)
feed_profile
```

```python
# Standardizing weight(X) data
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(feed_profile)
```

```python
# applying PCA
from sklearn.decomposition import PCA
pca = PCA(n_components=1)
X_pca = pca.fit_transform(X_scaled)

X_pca
```
